## Acerca del Data Set 
El conjunto de datos proporcionado es un archivo CSV que contiene información tabular con múltiples columnas y filas.
Cada Columna representa una variable diferente, y cada Fila representa un registro diferente. Algunas columnas pueden contener datos numéricos, mientras que otras pueden contener datos categóricos o de texto. El objetivo es limpiar y preparar estos datos para su análisis posterior.


**Amount:** Representa una cantidad monetaria del reclamo de los familiares en dolares


**Severity:** Es la calificación de gravedad del daño al paciente, de 1 (trauma emocional) a 9 (muerte)

**Nivel,Descripción**


    9: Muerte


    8: Daño permanente grave


    7: Daño permanente


    6: Daño temporal


    5: Tratamiento adicional


    4: Angustia o incomodidad emocional


    3:"No hay daño evidente, físico o de otro tipo."


    2: No llegó al paciente

    
    1: Condición insegura


**Age:** Edad del reclamante en años


**Private Attorney:** En caso de que el reclamante haya sido representado por algun abogado privado

    0: No ha sido representado
    1: Fue representado

**Marital status:** Estado civil del reclamante


Estado civil: 0 (Divorciado), 1 (Soltero), 2 (Casado), 3 (Viudo), 4 (Desconocido)

**Speciality:** Especialidad del medico involucrado en la demanda


**Insurance:** Tipo de Seguro medico con el que contaba el paciente

**Medicare** es principalmente para personas mayores de 65 años y para personas con ciertas discapacidades, independientemente de sus ingresos. **Medicaid**es un programa estatal y federal que proporciona cobertura médica a personas de bajos ingresos, familias y niños, mujeres embarazadas y personas con discapacidades


**Género:** Género del Paciente afectado

## Moda


Las ultimas 3 columnas de la base de datos sera rellenada con la moda ya que considero es lo mas conveniente. 

```python
moda_specialty = df2['Especialidad'].mode()[0]
df2['Especialidad'] = df2['Especialidad'].fillna(moda_specialty)
```

```python
moda_insurance = df2['Seguro'].mode()[0]
df2['Seguro'] = df2['Seguro'].fillna(moda_insurance)
```

```python
moda_gender = df2['Género'].mode()[0]
df2['Género'] = df2['Género'].fillna(moda_gender)
```

```python
# 1. Mediana de la columna 'Importe'
mediana_importe = df2['Monto'].median()

# 2. Mediana de la columna 'Edad'
mediana_edad = df2['Edad'].median()

# 3. Imprimir los resultados
print(f"Mediana del Monto: {mediana_importe:.2f}")
print(f"Mediana de la Edad: {mediana_edad:.2f}")
```

```python
# Columna de Importe
desviacion_importe = df2['Monto'].std()
min_importe = df2['Monto'].min()
max_importe = df2['Monto'].max()

# Columna de Edad
desviacion_edad = df2['Edad'].std()
min_edad = df2['Edad'].min()
max_edad = df2['Edad'].max()

print("--- Análisis de Dispersión y Rango ---")

## 1. Importe
print("\n Importe:")
print(f"   - Desviación Estándar (STD): {desviacion_importe:.2f}")
print(f"   - Valor Mínimo (Min): {min_importe:.2f}")
print(f"   - Valor Máximo (Max): {max_importe:.2f}")

## 2. Edad
print("\n Edad:")
print(f"   - Desviación Estándar (STD): {desviacion_edad:.2f}")
print(f"   - Valor Mínimo (Min): {min_edad:.2f}")
print(f"   - Valor Máximo (Max): {max_edad:.2f}")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Asumiendo que tu DataFrame limpio se llama df

# 1. Definir los rangos de edad (Bins)
# Definimos los límites para agrupar las edades, por ejemplo: 0-20, 20-40, 40-60, 60-80, 80+
bins = [0, 20, 40, 60, 80, df['Edad'].max() + 1] 
labels = ['0-20', '21-40', '41-60', '61-80', '80+']

# 2. Crear la nueva columna categórica 'Rango_Edad'
# Nota: La columna 'Monto' debe ser la que tiene los valores de Importe/Liquidación.
df['Rango_Edad'] = pd.cut(df['Edad'], bins=bins, labels=labels, right=False) 

# 3. Crear el Box Plot (Rango de Edad vs. Monto)
plt.figure(figsize=(10, 6))

sns.boxplot(
    x='Rango_Edad',      # Eje Horizontal (Rangos de Edad)
    y='Monto',           # Eje Vertical (Monto de la Liquidación)
    data=df,
    palette='viridis'    # Paleta de colores atractiva
)

plt.title('Distribución del Monto de Liquidación por Rangos de Edad', fontsize=14)
plt.xlabel('Rango de Edad del Paciente', fontsize=12)
plt.ylabel('Monto de Liquidación', fontsize=12)
plt.grid(axis='y', linestyle='--') # Agregar líneas de cuadrícula para mejor lectura
plt.show()
```

```python
import pandas as pd
import numpy as np

# --- PASO 1: DEFINICIÓN DE LA COLUMNA Y CÁLCULO DEL IQR ---

columna_outliers = 'Monto' 
# Si tu columna aún se llama 'Monto', cambia la línea anterior a: columna_outliers = 'Monto'

# 1. Calcular el primer cuartil (Q1) y el tercer cuartil (Q3)
Q1 = df1[columna_outliers].quantile(0.25)
Q3 = df1[columna_outliers].quantile(0.75)
IQR = Q3 - Q1

# Definir los límites de la valla (1.5 veces el IQR)
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print(f"Cuartil 1 (Q1): {Q1:.2f}")
print(f"Cuartil 3 (Q3): {Q3:.2f}")
print(f"Límite Superior para Outliers: {limite_superior:.2f}")
print("-" * 40)

# --- PASO 2: ELIMINACIÓN Y ACTUALIZACIÓN DEL DATAFRAME ---

# 1. Identificar los outliers
outliers = df1[
    (df1[columna_outliers] < limite_inferior) | (df1[columna_outliers] > limite_superior)
]

# 2. Reemplazar el DataFrame 'df' con la versión sin los outliers
df1 = df1[
    (df1[columna_outliers] >= limite_inferior) & (df1[columna_outliers] <= limite_superior)
].copy()

# 3. Reportar el resultado
print(f"Registros originales antes del filtro: {len(outliers) + len(df1)}")
print(f"Registros eliminados (Outliers de Importe): {len(outliers)}")
print(f"Registros finales en df: {len(df1)}")
```

## Modelo de Machine Learning

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import numpy as np
```

```python
X = df.drop('Especialidad', axis=1)
y = df['Especialidad']
# Utilizamos LabelEncoder para convertir las etiquetas de texto a números (ej. Cardiología -> 0, Pediatría -> 1)
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Usamos One-Hot Encoding para evitar que el modelo interprete las categorías como un orden (ej. 1 > 0).
# Identificar columnas categóricas restantes (usando los nombres traducidos)
columnas_categoricas = ['Estado Civil', 'Seguro', 'Género'] # Excluye Especialidad y las numéricas
X_encoded = pd.get_dummies(X, columns=columnas_categoricas, drop_first=True)
print("Variables codificadas y listas para el entrenamiento.")

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

print(f"Tamaño del conjunto de entrenamiento (X_train): {X_train.shape}")
print(f"Tamaño del conjunto de prueba (X_test): {X_test.shape}")

# Inicializar y entrenar el modelo Random Forest
rf_model = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)

print("Iniciando entrenamiento...")
rf_model.fit(X_train, y_train)
print("Entrenamiento completado.")
```

```python
# Cálculo que ya está en tu código:
accuracy = accuracy_score(y_test, y_test)
# Ejemplo: 0.85
from sklearn.metrics import classification_report
# ... (asegúrate de que y_test y y_pred estén codificados numéricamente)

# Descodificar las etiquetas para el reporte
nombres_clases = le.classes_
print(classification_report(y_test, y_test, target_names=nombres_clases, zero_division=0))
```
